## Interpretación de resultados

El análisis de priorización se realizó utilizando las predicciones finales del modelo de bosque, con un total de 713 clientes pertenecientes al conjunto de prueba. El objetivo es identificar cuáles presentan mayor riesgo de abandono y orientar las acciones de retención según la capacidad de contacto de la empresa.

Los resultados muestran que el primer decil concentra 67 de los 112 abandonos reales, lo que representa una tasa de abandono del 93,1 %. Al ampliar la selección al 20 % de los clientes con mayor riesgo, es decir, 143 clientes, se logran identificar 102 abandonos, equivalentes al 91,1 % del total.

En cuanto a la capacidad de contacto, los resultados indican que la empresa puede capturar el 58,9 % de los abandonos contactando al 10 % de los clientes, el 91,1 % contactando al 20 % y el 96,4 % contactando al 30 %. Esto permite evaluar diferentes estrategias según los recursos disponibles.

Al comparar el 20 % de clientes con mayor riesgo frente al resto, se observa que el grupo prioritario presenta menores medianas de segundos de uso, valor del cliente, frecuencia de llamadas, números distintos y frecuencia de SMS. Estas diferencias son estadísticamente significativas. Por otro lado, los meses de suscripción y las fallas de llamada no muestran diferencias significativas.

En las variables categóricas se destaca que el 89,1 % de los clientes que presentaron quejas y el 68,5 % de los clientes con estado 2 quedaron dentro del grupo prioritario. Estos resultados ayudan a describir las características de los clientes identificados por el modelo, aunque no implican que dichas características causen el abandono.

Finalmente, el análisis de costos muestra que el umbral de decisión más conveniente depende de la importancia que la empresa asigne a evitar falsos negativos. Cuando el costo de no detectar un abandono aumenta, resulta conveniente considerar umbrales más bajos para identificar una mayor cantidad de clientes en riesgo.

En conjunto, los resultados muestran que el modelo de bosque permite ordenar los clientes según su riesgo de abandono y apoyar la asignación de los recursos de retención. Estas predicciones deben entenderse como una herramienta de apoyo a la decisión y no como una garantía de que un cliente abandonará el servicio.
